## Language Model
LangChain의 Intergrations 섹션에서는 다양한 다운스트림 LLM 모델과의 연동을 지원한다.

### openai

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model_name='gpt-4.1-mini', temperature=0.9)

llm.invoke("태국의 수도는 어디인가요?")
```

```python
# 추상화 된 lm 모델 객체 생성
from langchain.chat_models import init_chat_model

# llm = init_chat_model(model='gpt-4.1-mini', model_provider='openai', temperature=0.9)
llm = init_chat_model("openai:gpt-4.1-mini", temperature=0.9)
ai_message = llm.invoke('핀란드의 여행지를 추천해주세요.')
print(ai_message.content)
```

## huggingface

```python
from langchain.chat_models import init_chat_model

llm = init_chat_model("huggingface:Qwen/Qwen2.5-0.5B-Instruct", temperature=0.9)
ai_message = llm.invoke('Translate this to French: I love programming.')
print(ai_message.content)
```

```python
# 여러 LLM을 동시에 비교
models = [
    init_chat_model('openai:gpt-4.1-mini'),
    init_chat_model('huggingface:Qwen/Qwen2.5-0.5B-Instruct')
]

user_prompt = "LangChain의 장점은 뭘까?"

for model in models:
    ai_message = model.invoke(user_prompt)
    print(f"Responses : {ai_message.content}\n")
```

## Prompts

```python
llm = init_chat_model("openai:gpt-4.1-mini")

message = [
    ('system', '당신은 친절한 초딩 전용 챗봇입니다. 초딩의 눈높이에 맞게 설명해주세요.'),
    ('user', '랭체인이 뭔가요?')
]

ai_message = llm.invoke(message)
print(ai_message.content)
```

```python
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage, ToolMessage

message = [
    SystemMessage(content="당신은 친절한 초딩 전용 챗봇입니다. 초딩의 눈높이에 맞게 설명해주세요."),
    HumanMessage(content="랭체인이 뭔가요?")
]

ai_message = llm.invoke(message)
print(type(ai_message))
print(ai_message.content)
```

### PromptTemplate

```python
from langchain_core.prompts import PromptTemplate

# 특정 상품에 대한 광고 문구를 생성
prompt = PromptTemplate.from_template('{product}를 홍보하기 위한 신박한 광고 문구를 작성해줘.')
print(prompt.format(product='전기차'))
print(prompt.format(product='카메라'))

ai_message = llm.invoke(prompt.format(product='카메라'))
print(ai_message.content)
```

### ChatPromptTemplate

```python
from langchain_core.prompts import ChatPromptTemplate

chat_prompt = ChatPromptTemplate.from_messages([
    ('system', '당신은 {domain} 분야의 전문가입니다. 해당 분야에 대한 질문에 친절히 답해주세요.'),
    ('user', '{question}\n한글로 답해주세요.')
])

print(chat_prompt)
print(chat_prompt.format(domain='AI', question='랭스미스 사용법'))

ai_message = llm.invoke(
    chat_prompt.format(domain='AI', question='랭스미스 사용법')
)

print(ai_message.content)
```

## Output Parsers
- LLM이 생성한 텍스트 출력을 특정 형식으로 변환하거나 처리

### CommaSeparatedListOutputParser

```python
from langchain_core.output_parsers import CommaSeparatedListOutputParser

model_output = "사과, 바나나, 오렌지, 포도"

output_parser = CommaSeparatedListOutputParser()
output = output_parser.parse(model_output)
output
```

```python
format_instruction = output_parser.get_format_instructions()
print(format_instruction)
```

```python
prompt_template = PromptTemplate(
    template="{subject} {n}개의 팀을 보여주세요.\n{format_instruction}",
    input_variables=["subject", "n"],   # 사용자 프롬프트로 채워질 변수 
    partial_variables={
        # template 생성 시에 채워질 변수
        'format_instruction' : output_parser.get_format_instructions()
    }
)

prompt = prompt_template.format(subject="대한민국 프로야구", n=5)
print(prompt)
```

```python
ai_message = llm.invoke(prompt)
output = ai_message.content
print('모델 출력 : ', output)

output = output_parser.parse(output)
print('최종 출력 : ', output)
```

```python
# LangChain 의 LCEL(LangChain Expression Language) 방식
chain = prompt_template | llm | output_parser
result = chain.invoke(dict(subject='프리미어리그', n=5))
print('최종 결과 : ', result)
```

### StrOutputParser
- 모델 응답 텍스트를 그대로 반환

```python
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()
chain = llm | parser
print(chain.invoke("봄 제철 과일은 뭐가 있어?"))
```

### JsonOutputParser
모델의 json 문자열 응답을 파이썬 객체 dict/list 로 변환

```python
from langchain_core.output_parsers import JsonOutputParser

json_parser = JsonOutputParser()
print(json_parser.get_format_instructions())

model_output = '{"title":"GPT-5를 소개합니다", "author":"OpenAI", "pages":250}'
output = json_parser.parse(model_output)    # JSON str -> python object(dict or list)
print(output)
print(type(output))
```

```python
prompt_template = PromptTemplate(
    template='{subject} 분야의 유명한 한국어 책 {n}권의 정보를 제공하세요.\n{format_instruction}',
    input_variables=['subject', 'n'],
    partial_variables={
        'format_instruction' : json_parser.get_format_instructions()
    }
)

prompt = prompt_template.format(subject='인공지능', n=3)
ai_message = llm.invoke(prompt)
output = json_parser.parse(ai_message.content)
print(output)
```

```python
chain = prompt_template | llm | json_parser
chain.invoke(input={'subject':'인공지능', 'n':3})
```

```python
# 위의 예제에서는 JSON 속성 값을 고정하지 않아 호출 시 마다 계속 변경 된다
# PydanticOutputParser : 모델 출력이 특정 Pydantic 모델의 형태를 갖추도록 강제하는 출력 파서
from typing import List
from pydantic import BaseModel
from langchain_core.output_parsers import PydanticOutputParser

class BookInfo(BaseModel):
    title: str
    author: str
    publisher: str
    year: int
    description: str

class BookInfoList(BaseModel):
    books: List[BookInfo]

parser = PydanticOutputParser(pydantic_object=BookInfoList)

prompt_template = PromptTemplate(
    template='{subject} 분야의 유명한 한국어 책 {n}권의 정보를 제공하세요.\n{format_instruction}',
    input_variables=['subject', 'n'],
    partial_variables={
        'format_instruction' : parser.get_format_instructions()
    }
)

chain = prompt_template | llm | parser

print(chain.invoke({'subject': '인공지능', 'n': 3}))
```

```python
# 간소화 버전
prompt = PromptTemplate(
    template='{subject} 분야의 유명한 한국어 책 {n}권의 정보를 제공하세요.',
    input_variables=['subject', 'n']
)

chain = prompt | llm.with_structured_output(BookInfoList)

print(chain.invoke({'subject': '인공지능', 'n': 3}))
```
